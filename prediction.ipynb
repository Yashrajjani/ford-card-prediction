```python
x = df.drop(columns=['price'])
y = df['price']
```

```python
print(x)
```

```python
df.columns
```

```python
x_one_hot_encode = pd.get_dummies(x,columns=['model','transmission','fuelType'],drop_first=True)
x_one_hot_encode = x_one_hot_encode.astype(int)
x_one_hot_encode
```

```python
from sklearn.preprocessing import LabelEncoder

columns = ['model', 'transmission', 'fuelType']

Xlable = x.copy()  # make a safe copy
label_encoders = {}

for col in columns:
    le = LabelEncoder()
    Xlable[col] = le.fit_transform(Xlable[col].astype(str))  # Convert to string in case of nulls
    label_encoders[col] = le

Xlable
```

```python
from sklearn.preprocessing import StandardScaler
numerical_cols = ['year', 'mileage', 'tax', 'mpg', 'engineSize']
scaler = StandardScaler()
x_one_hot_encode[numerical_cols] = scaler.fit_transform(x_one_hot_encode[numerical_cols])
```

```python
x_one_hot_encode
```

```python
from sklearn.preprocessing import StandardScaler
numerical_cols = ['year', 'mileage', 'tax', 'mpg', 'engineSize']
scaler = StandardScaler()
Xlable[numerical_cols] = scaler.fit_transform(Xlable[numerical_cols])
```

```python
Xlable
```

```python
Xlable[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']] = scaler.fit_transform(Xlable[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']])

Xlable
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

X_train, X_test, y_train, y_test = train_test_split(x_one_hot_encode, y, test_size=0.20,random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import r2_score
r2 = r2_score(y_test,y_pred)
r2
```

```python
n = X_test.shape[0]
p = X_test.shape[1]
adjusted_r2 = 1 - ((1 - r2) * (n - 1)) / (n - p - 1)
print("Adjusted R² Score:", adjusted_r2)
```
